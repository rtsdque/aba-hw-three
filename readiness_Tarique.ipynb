VS Code seems to be better for projects with multiple moving pieces, such as if you were designing a quantitative trading model, where you had one file containing a historical backtest of your strategy, and another file containing a predicted future test of your strategy. VS Code would also allow you to easily have several different file types in one places, so you could also keep a folder with all your historical data right in that same VS Code session.

Jupyter is better for tasks with less moving pieces, and a more focused workflow in mind. This can be great if you have, for example, some medical data revolving around a new drug, where you can import that data and then run multiple tests, in a top down format, where you can continue with testing as each previous block comes back with the desired response.
